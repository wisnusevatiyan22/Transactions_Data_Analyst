```python
import pandas as pd
import sqlite3

# Membaca file Excel dengan nama kolom hanya dari tabel 1
file1 = pd.read_excel('01. table_transaksi_jan_2020.xlsx')
file2 = pd.read_excel('02. table_transaksi_feb_2020.xlsx', header=None)  # Mengabaikan nama kolom
file3 = pd.read_excel('03. table_transaksi_mar_2020.xlsx', header=None)  # Mengabaikan nama kolom

# Hanya menggunakan data dari file 1 untuk nama kolom
file2.columns = file1.columns  # Mengganti nama kolom tabel 2 dengan nama kolom tabel 1
file3.columns = file1.columns  # Mengganti nama kolom tabel 3 dengan nama kolom tabel 1

# Koneksi ke SQLite database (misalnya 'transaksi.db')
conn = sqlite3.connect('transaksi.db')  # Ganti dengan nama file .db yang diinginkan

# Menyimpan data dari Excel ke tabel SQL
file1.to_sql('table1', conn, if_exists='replace', index=False)
file2.to_sql('table2', conn, if_exists='replace', index=False)
file3.to_sql('table3', conn, if_exists='replace', index=False)

# SQL query untuk menggabungkan ketiga tabel
query = """
    SELECT * FROM table1
    UNION ALL
    SELECT * FROM table2
    UNION ALL
    SELECT * FROM table3
"""

# Menjalankan query dan memuat hasilnya ke dalam DataFrame
result = pd.read_sql_query(query, conn)

# Menyimpan hasil penggabungan ke tabel baru di SQLite
result.to_sql('transaksi_q1', conn, if_exists='replace', index=False)

print("Penggabungan selesai, hasil disimpan di tabel 'transaksi_q1' dalam SQLite database.")

# Menutup koneksi
conn.close()
```

```python
import pandas as pd

# Konversi kolom ke tipe data yang sesuai jika belum dilakukan
transaksi_Q1['sd_price'] = pd.to_numeric(transaksi_Q1['sd_price'], errors='coerce')
transaksi_Q1['rs_price'] = pd.to_numeric(transaksi_Q1['rs_price'], errors='coerce')

# Hapus baris dengan NaN jika ada setelah konversi
transaksi_Q1.dropna(subset=['sd_price', 'rs_price'], inplace=True)

# Hitung profit margin untuk setiap transaksi
transaksi_Q1['profit_margin'] = transaksi_Q1['rs_price'] - transaksi_Q1['sd_price']

# Tentukan kriteria untuk kategori produk yang berbeda
kriteria_produk = [
    "Tagihan BPJS Bukan Penerima Upah",
    "Tagihan BPJS Penerima Upah",
    "Bayar BPJS Ketenagakerjaan BPU 2 Bulan",
    "Bayar BPJS Ketenagakerjaan BPU 3 Bulan"
]

# Buat kolom untuk nett profit
def hitung_nett_profit(row):
    if row['description'] in kriteria_produk:
        return 0.30 * row['profit_margin']  # 30% untuk produk tertentu
    else:
        return 0.50 * row['profit_margin']  # 50% untuk produk lainnya

transaksi_Q1['nett_profit'] = transaksi_Q1.apply(hitung_nett_profit, axis=1)

# Hitung total nett profit
total_nett_profit = transaksi_Q1['nett_profit'].sum()
print(f"Total nett profit selama Q1 2020: {total_nett_profit}")
```
